```python
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('rata-rata harga:', harga.mean())
print('harga tertinggi:', harga.max())
print('harga setelah diskon 10%', harga * 0.9)
```

analisis 1

Jika harga berupa **list Python biasa** seperti [5000, 7000, 3000, 12000, 4500] lalu dikalikan 0.9,
maka akan muncul **TypeError**, karena operator `*` pada list Python hanya bisa digunakan dengan
bilangan bulat (int) untuk mengulang/menduplikasi isi list, bukan untuk mengalikan tiap elemennya.
Karena 0.9 adalah bilangan desimal (float), Python tidak tahu cara "mengulang" list dengan angka
pecahan

```python
import pandas as pd

data_kantin = {
    'menu': ['nasi goreng', 'es teh', 'mie ayam', 'es teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

analisis 2

kolom yang memiliki data kosong adalah menu dan terjual, jika data langsung dianalisis dan tidak dibersihkan, maka hasil analisis akan tidak akurat

```python
df = pd.read_csv('data_kantin - data_kantin.csv')

print(df.head())
print(df.info())
print(df.describe())
print(df.shape)
```

analisis 3

kolom yang jumlah non-null yang lebih sedikit adalah menu dan terjual, berarti terdapat data yang kosong pada kolom menu dan terjual




```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom

df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

analisis 4

kolom terjual diisi (fillna) karena isi kolom terjual berupa *integer* yang dimana data yang kosong dapat dicari dengan nilai yang ditentukan, lalu kolom menu diisi (dropna) karena isi kolom menu berupa *string* yang membuat nilai yang kosong tidak dapat dicari dengan nilai yang ditentukan, oleh karena itu lebih aman dengan menghapus data dari kolom menu yang kosong


```python
print(df.duplicated().sum())
df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int)
print (df.dtypes)
```

Dataset berkurang dari 107 baris → 102 baris, memastikan dtypes benar sebelum analisis itu penting supaya operasi berjalan sesuai yang diharapkan dan hasil analisis tidak salah atau error di tengah jalan

```python
laris = df[df['terjual'] > 20]
urut = df.sort_values(by='terjual', ascending=False)

df['total_pendapatan'] = df['harga'] * df['terjual']

ringkasan = df.groupby('menu')['total_pendapatan'].sum()
print(ringkasan)
```
