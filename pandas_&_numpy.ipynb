```python
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('rata-rata harga:', harga.mean())
print('harga tertinggi:', harga.max())
print('harga setelah diskon 10%', harga * 0.9)
```

analisis 1

Jika harga berupa **list Python biasa** seperti [5000, 7000, 3000, 12000, 4500] lalu dikalikan 0.9,
maka akan muncul **TypeError**, karena operator `*` pada list Python hanya bisa digunakan dengan
bilangan bulat (int) untuk mengulang/menduplikasi isi list, bukan untuk mengalikan tiap elemennya.
Karena 0.9 adalah bilangan desimal (float), Python tidak tahu cara "mengulang" list dengan angka
pecahan

```python
import pandas as pd

data_kantin = {
    'menu': ['nasi goreng', 'es teh', 'mie ayam', 'es teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

analisis 2

kolom yang memiliki data kosong adalah menu dan terjual, jika data langsung dianalisis dan tidak dibersihkan, maka hasil analisis akan tidak akurat

```python
df = pd.read_csv('data_kantin - data_kantin.csv')

print(df.head())
print(df.info())
print(df.describe())
print(df.shape)
```

analisis 3

kolom yang jumlah non-null yang lebih sedikit adalah menu dan terjual, berarti terdapat data yang kosong pada kolom menu dan terjual




```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom

df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

analisis 4

kolom terjual diisi (fillna) karena isi kolom terjual berupa *integer* yang dimana data yang kosong dapat dicari dengan nilai yang ditentukan, lalu kolom menu diisi (dropna) karena isi kolom menu berupa *string* yang membuat nilai yang kosong tidak dapat dicari dengan nilai yang ditentukan, oleh karena itu lebih aman dengan menghapus data dari kolom menu yang kosong


```python
print(df.duplicated().sum())
df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int)
print (df.dtypes)
```

```python
df = pd.read_csv('dataset_penjualan_kantin - dataset_penjualan_kantin.csv')

print(df.head())
print(df.info())
print(df.describe())
print(df.shape)
```

```python
def bersihkan_tanggal(teks):
    teks = str(teks).strip()
    for fmt in ("%Y-%m-%d", "%d/%m/%Y", "%d %B %Y"):
        hasil = pd.to_datetime(teks, format=fmt, errors="coerce")
        if pd.notna(hasil):
            return hasil
    return pd.NaT

df["tanggal"] = df["tanggal"].apply(bersihkan_tanggal)
print(df["tanggal"].head())
```

```python
df['nama_produk'] = (df['nama_produk'].str.lower().str.title())
print(df["nama_produk"])
```

```python
if df["jumlah_terjual"].dtype == object:
    df["jumlah_terjual"] = df["jumlah_terjual"].astype(str)
    df["jumlah_terjual"] = df["jumlah_terjual"].str.replace(r"[^0-9]", "", regex=True)
    df["jumlah_terjual"] = df["jumlah_terjual"].replace("", np.nan)

df["jumlah_terjual"] = pd.to_numeric(df["jumlah_terjual"])
print(df["jumlah_terjual"].head())
```

```python
df["harga_satuan"] = df["harga_satuan"].astype(str)
df["harga_satuan"] = df["harga_satuan"].str.replace(r"[^0-9]", "", regex=True)
df["harga_satuan"] = df["harga_satuan"].replace("", np.nan)
df["harga_satuan"] = pd.to_numeric(df["harga_satuan"])
print(df["harga_satuan"].head())
```

```python
print(df.isnull().sum())
df['jumlah_terjual'] = df['jumlah_terjual'].fillna(0)
df['nama_kasir'] = df['nama_kasir'].fillna('tidak diketahui')
print(df[df["jumlah_terjual"] > 100])
df = df[df["jumlah_terjual"] <= 100]
```

```python
print(df.duplicated(subset=["id_transaksi"]).sum())
df = df.drop_duplicates(subset=["id_transaksi"])

print(df.dtypes)
```

```python
df["harga_satuan"] = df.groupby("nama_produk")["harga_satuan"].transform(
    lambda x: x.fillna(x.median())
)
print(df.info)
```

```python
laku = df[df['jumlah_terjual'] > 20]
urut = df.sort_values(by='jumlah_terjual', ascending=False)

df['total_pendapatan'] = df['harga_satuan'] * df['jumlah_terjual']

ringkasan = df.groupby('nama_produk')['total_pendapatan'].sum()
print(ringkasan)

metode_terbanyak = df['metode_pembayaran'].value_counts()
print(metode_terbanyak)
```

kami memilih menggunakan df.fillna(0) karena untuk kolom jumlah terjual bisa juga di setting 0 atau artinya belum ada data , lalu untuk data kosong pada kolom nama kasir kami buat data kosong untuk terisi "tidak diketahui" karena menurut kami lebih bagus diisi daripada di hapus . dan yang terakhir untuk kolom harga satuan kami memilih mengisi harga yang kosong dengan harga yang sama sesuai menu nya
